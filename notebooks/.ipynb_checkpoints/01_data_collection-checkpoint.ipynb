```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df['Date'] = pd.to_datetime(df['Date'])

fig, ax = plt.subplots(figsize=(14,6))

ax.plot(df['Date'], df['Close'],
        color='blue',
        linewidth=2,
        label='Close Price')

ax.plot(df['Date'], df['Open'],
        color='red',
        linestyle='--',
        linewidth=2,
        label='Open Price')

ax.xaxis.set_major_locator(mdates.YearLocator(1))      
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

plt.xticks(rotation=45)

plt.xlabel("Year")
plt.ylabel("Price (USD)")
plt.title("Bitcoin Open vs Close Price")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df=df.drop(['Adj Close'], axis=1)
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year

yearly_volume = df.groupby('Year')['Volume'].sum()
plt.figure(figsize=(8,8))

plt.pie(
    yearly_volume,
    labels=yearly_volume.index,
    autopct='%1.1f%%'
)

plt.title("Bitcoin Trading Volume Distribution by Year")

plt.legend(
    title="Year",
    loc="center left",
     bbox_to_anchor=(1, 0.5)
   
)

plt.tight_layout()
plt.show()
```

```python
df['Date']=pd.to_datetime(df['Date'])
df['Year']=df['Date'].dt.year
yearly_volume=df.groupby('Year')['Volume'].sum()
print(yearly_volume)
```

```python
import seaborn as sn
features = ['Open', 'High', 'Low', 'Close']

plt.subplots(figsize=(20,10))
for i, col in enumerate(features):
  plt.subplot(2,2,i+1)
  sn.distplot(df[col])
plt.show()
```

```python
df['Date']=pd.to_datetime(df['Date'])
df['year']=df['Date'].dt.year
df['month']=df['Date'].dt.month
df['day']=df['Date'].dt.day
df.head()
```

```python
sns.boxplot(df['Close'])
```

```python
yearly_median=df.groupby('Year')['Close'].median()
```

```python
print(yearly_median)
```

```python
data_grouped = df.groupby('year').mean()

plt.subplots(figsize=(20, 10))

for i, col in enumerate(['Open', 'High', 'Low', 'Close']):
    plt.subplot(2, 2, i + 1)
    data_grouped[col].plot.bar()
    plt.title(f'Year-wise Average {col} Price')
    plt.xlabel('Year')
    plt.ylabel('Price')

plt.tight_layout()
plt.show()
```

```python
df['is_quarter_end'] = np.where(df['month']%3==0,1,0)
df.head()
```

```python
df['price_change'] = df['Close']-df['Open']
```

```python
df['High_low_range']=df['High']-df['Low']
```

```python
#this tells the relative position of todays price
df['close_position'] = (
    (df['Close'] - df['Low']) /
    (df['High'] - df['Low'])
)
```
